```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
X = np.random.random((100,10))
```

```python
#(1, 10)
#(100, 10)
```

```python
%%time
def mnn(X, m):
    n = len(X)
    S = np.zeros((n, m))
    for i in range(n):
        distances = np.zeros_like(X)
        distances = np.sum((X[i] - X)**2, axis=1)
        
        dist_ind = np.argsort(np.r_[distances[:i], distances[i+1:]])[:m] + 1
        S[i, :] = dist_ind
    return S
mnn(X, 5)
```

```python
def vectorized_mnn(X, m):
    dot_product = np.dot(X, X.T)
    a_ = np.diagonal(dot_product) #the same is b_ beacuse symmetric matrix
    distances = a_[:, np.newaxis] - 2*dot_product + a_[np.newaxis, :]
    
    sorted_arr = np.argpartition(distances, m+1, axis=1)[:, 1:m+1]
    
    n = len(sorted_arr)
    final_indices = np.zeros((n,m))
    for i in range(n):
        row = sorted_arr[i][sorted_arr[i] != i]
        final_indices[i] = row[:]
    return final_indices
vectorized_mnn(X, 3)
```
