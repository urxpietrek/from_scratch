```python
import numpy as np
```

```python
np.random.seed(37)
n_samples = 78
n_bootstrap = 1000
x = np.random.randn(n_samples) + np.random.random(n_samples) * np.e
bootstrapped_x = np.random.choice(x, size = (n_bootstrap, n_samples), replace=True)
means = np.mean(bootstrapped_x, axis=1)
import matplotlib.pyplot as plt
plt.hist(means, bins=30)
sg_lvl = 0.05

for p in (np.percentile(means, q=[sg_lvl/2 * 100, (1-sg_lvl/2)*100])):
    plt.axvline(p)
```
