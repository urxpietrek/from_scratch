```python
import matplotlib.pyplot as plt
def K(x):
    return np.exp(-np.power(x, 2)/2)/math.sqrt(2*math.pi)

def f(X, h):
    x_grid = np.linspace(np.min(X), np.max(X),)
    n = X.size
    total_sum = np.sum(K((x_grid - X[:, np.newaxis])/h), axis=0)
    """total_sum = 0
    for xi in X:
        total_sum += K((x_grid - xi)/h)"""
    y_range = total_sum/(h*n)
    plt.plot(x_grid, y_range,  
            label=f'{h}')

    plt.plot(data, np.zeros_like(data) , 's', 
            markersize=8, color='black')
        
    return total_sum
f(data, h)
```
